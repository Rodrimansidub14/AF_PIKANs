## Puntos generados: Allen–Cahn, entrenamiento vs. held-out independiente

```python
import numpy as np, pandas as pd, plotly.express as px
pools = np.load(BASELINE_RUNS / "seed-101" / "allen_cahn" / "pools.npz")
training = pools['interior']; heldout = pools['heldout']
points = pd.DataFrame({'x': np.r_[training[::16,0], heldout[::32,0]],
    't': np.r_[training[::16,1], heldout[::32,1]],
    'pool': ['training']*len(training[::16]) + ['held-out diagnostic']*len(heldout[::32])})
figure = px.scatter(points, x='x', y='t', color='pool', opacity=.55,
    labels={'x':'space coordinate x','t':'time coordinate t'},
    title='Allen–Cahn independent training and held-out Sobol pools, seed 101')
figure
```
